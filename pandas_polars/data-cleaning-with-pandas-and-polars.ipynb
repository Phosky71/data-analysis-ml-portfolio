### 3.2 Cargar con Polars
Realiza la misma operación utilizando Polars.

```python
# Tu código aquí (Polars)
df_polars = pl.read_csv('netflix_titles.csv') # Ajusta el nombre del archivo
df_polars.head()
```

## 4. Inspección General del Dataset

## 5. Manejo de Valores Faltantes (Nulls/NaNs)

Los datos reales raramente están completos.
**Tarea:**
1.  Cuenta cuántos valores nulos hay por columna.
2.  Decide una estrategia para una columna importante con nulos (e.g., rellenar con la media, moda, o un valor fijo como "Desconocido", o eliminar las filas).
3.  Aplica esa estrategia.

```python
# --- PANDAS ---
print("Nulos por columna:")
print(df_pandas.isnull().sum())

cols_fillna = ['director', 'cast', 'country']
df_pandas[cols_fillna] = df_pandas[cols_fillna].fillna("Unknown")

df_pandas = df_pandas.dropna(subset=['date_added'])

print("\nNulos tras limpieza:")
print(df_pandas.isnull().sum())
```

```python
# --- POLARS ---
print("Nulos por columna:")
print(df_polars.null_count())

df_polars = df_polars.with_columns([
    pl.col("director").fill_null("Unknown"),
    pl.col("cast").fill_null("Unknown"),
    pl.col("country").fill_null("Unknown")
]).drop_nulls(subset=["date_added"])

print("\nNulos tras limpieza:")
print(df_polars.null_count())
```

## 6. Manejo de Duplicados

**Tarea:** Verifica si existen filas totalmente duplicadas y elimínalas.

```python
# --- PANDAS ---
# Verificar y eliminar duplicados
duplicados = df_pandas.duplicated(subset=['show_id']).sum()
print(f"Duplicados encontrados: {duplicados}")

df_pandas = df_pandas.drop_duplicates(subset=['show_id'])
```

```python
# --- POLARS ---
# Verificar y eliminar duplicados (unique)
n_dupes = df_polars.height - df_polars.unique(subset=["show_id"]).height
print(f"Duplicados encontrados: {n_dupes}")

df_polars = df_polars.unique(subset=["show_id"])
```

## 7. Transformación de Datos y Limpieza de Strings

A menudo, los números vienen como texto (e.g., "$1,000", "50 kg") o las fechas no tienen el formato correcto.

**Tarea (Ejemplo Genérico):**
1.  Selecciona una columna de tipo string que debería ser numérica o fecha.
2.  Limpia el string (quita símbolos como '$', 'kg', ',').
3.  Convierte la columna al tipo correcto (int, float, o datetime).

*Nota: Si elegiste el dataset de coches, la columna de 'precio' o 'kilometraje' es ideal para esto.*

```python
# --- PANDAS ---
df_pandas['date_added'] = df_pandas['date_added'].str.strip()
df_pandas['date_added_clean'] = pd.to_datetime(df_pandas['date_added'], errors='coerce')

df_pandas['duration_val'] = df_pandas['duration'].astype(str).str.extract(r'(\d+)').astype(float)

print(df_pandas[['date_added', 'date_added_clean', 'duration', 'duration_val']].head())
```

```python
# --- POLARS ---
df_polars = df_polars.with_columns([
    pl.col("date_added")
      .str.strip_chars()
      .str.to_date("%B %d, %Y", strict=False)
      .alias("date_added_clean"),

    pl.col("duration")
      .str.extract(r"(\d+)", 1)
      .cast(pl.Int64, strict=False)
      .alias("duration_val")
])

print(df_polars.select(['date_added', 'date_added_clean', 'duration', 'duration_val']).head())
```

Escribe brevemente tus conclusiones:

- ¿Notaste alguna diferencia de velocidad (si el dataset era grande)?
- ¿Qué dificultades encontraste en la limpieza?

En cuanto a la intuición, Polars me resultó más organizada para definir el flujo de limpieza; su sintaxis encadenada (.with_columns) hace que el código sea más legible que las múltiples reasignaciones de variables necesarias en Pandas. Sobre el rendimiento, aunque el dataset no es masivo, la ejecución en Polars se siente más optimizada, especialmente al convertir fechas, donde su enfoque estricto evita la sobrecarga de inferencia que hace Pandas.

La mayor dificultad fue limpiar columnas mixtas como duration y parsear date_added. Mientras que Pandas es más permisivo y "mágico", Polars obliga a ser explícito con los formatos de fecha y los tipos de datos. Esto añade una pequeña barrera inicial, pero a cambio ofrece un pipeline mucho más robusto y menos propenso a errores silenciosos.
