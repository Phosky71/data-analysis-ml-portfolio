```python
import string
import nltk
from nltk.tokenize import word_tokenize
import spacy

# Asegúrate de tener descargado el tokenizer de NLTK
# nltk.download('punkt')

# Cargar modelo de español de spaCy (debes haberlo instalado antes:
# python -m spacy download es_core_news_sm)
nlp = spacy.load("es_core_news_sm")

# 1. Pedir texto al usuario
texto = input("Introduce un texto: ")

# 2 y 3. Convertir a minúsculas y eliminar signos con la sentencia indicada
texto_limpio = texto.lower().translate(
    str.maketrans('', '', string.punctuation + "¡¿")
)

# 5 y 6. Tokenizar con NLTK
tokens_nltk = word_tokenize(texto_limpio)

# 7, 8 y 9. Tokenizar con spaCy
doc = nlp(texto_limpio)
tokens_spacy = [token.text for token in doc]

# 10. Pedir palabra que exista en el texto limpio
while True:
    palabra = input("Introduce una palabra que aparezca en el texto limpio: ").lower()

    if palabra in tokens_nltk:
        # Buscar la primera aparición y la siguiente palabra (si existe)
        indice = tokens_nltk.index(palabra)
        if indice + 1 < len(tokens_nltk):
            palabra_siguiente = tokens_nltk[indice + 1]
        else:
            palabra_siguiente = None
        break
    else:
        print("No existe la palabra, prueba otra vez.")

# 11. Mostrar resultados
print("\n--- RESULTADOS ---")
print("Texto limpio:")
print(texto_limpio)

print("\nTokens NLTK:")
print(tokens_nltk)

print("\nTokens spaCy:")
print(tokens_spacy)

print("\nPalabra introducida por el usuario:")
print(palabra)

print("\nPalabra que continúa a la introducida:")
print(palabra_siguiente if palabra_siguiente is not None else "No hay palabra siguiente.")
```
