```python
import re
import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import SnowballStemmer, WordNetLemmatizer

nltk.download('punkt_tab')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
# 1. Pedir al usuario que introduzca un párrafo en español
texto_original = input("Introduce un párrafo en español: ")

# 2. Limpieza básica del texto
# 2.1 Pasar a minúsculas
texto_minusculas = texto_original.lower()

# 2.2 Eliminar caracteres que no sean letras ni espacios
#    Dejamos letras (incluyendo acentos) y espacios
texto_solo_letras = re.sub(r"[^a-záéíóúüñ\s]", " ", texto_minusculas)

# 2.3 Normalizar espacios repetidos
texto_limpio = re.sub(r"\s+", " ", texto_solo_letras).strip()

# 3. Tokenización con nltk.word_tokenize
tokens = word_tokenize(texto_limpio, language="spanish")

# 4. Stemming con SnowballStemmer (español)
stemmer = SnowballStemmer("spanish")
tokens_stem = [stemmer.stem(tok) for tok in tokens]

# 5. Lematización con WordNetLemmatizer (inglés, pero se usa como práctica)
lemmatizer = WordNetLemmatizer()
tokens_lemma = [lemmatizer.lemmatize(tok) for tok in tokens]

# 6. Pedir una palabra del texto limpio y mostrar la siguiente si existe
palabra_usuario = ""
palabra_siguiente = None

while True:
    palabra_usuario = input("Introduce una palabra del texto limpio: ").lower().strip()

    if palabra_usuario in tokens:
        indice = tokens.index(palabra_usuario)
        if indice + 1 < len(tokens):
            palabra_siguiente = tokens[indice + 1]
        else:
            palabra_siguiente = None
        break
    else:
        print("La palabra no existe en el texto limpio. Inténtalo de nuevo.")

# 7. Mostrar por pantalla los resultados

print("\n--- RESULTADOS ---")
print("\nTexto original:")
print(texto_original)

print("\nTexto limpio:")
print(texto_limpio)

print("\nLista de tokens:")
print(tokens)

print("\nTokens con su stem y su lema:")
for tok, st, lm in zip(tokens, tokens_stem, tokens_lemma):
    print(f"Token: {tok:15} | Stem: {st:15} | Lema: {lm:15}")

print("\nResultado de la predicción (palabra siguiente):")
if palabra_siguiente is not None:
    print(f"La palabra siguiente a '{palabra_usuario}' es: '{palabra_siguiente}'")
else:
    print(f"La palabra '{palabra_usuario}' es la última y no tiene palabra siguiente.")
```
