# Introducción a las Series Temporales

En este cuaderno, exploraremos los conceptos básicos de las series temporales, su importancia en el análisis de datos y cómo visualizarlas.

## 1. ¿Qué es una Serie Temporal?

Una **serie temporal** es una secuencia de observaciones recogidas en intervalos regulares a lo largo del tiempo. Estas observaciones pueden ser cualquier tipo de dato que varíe con el tiempo, como:
- Datos climáticos (temperatura, precipitación).
- Datos económicos (ventas mensuales, precios de acciones).
- Datos de tráfico web.
- Datos de sensores en dispositivos IoT.

### Características de las Series Temporales
1. **Tendencia:** Cambio gradual a largo plazo (ej. aumento de temperatura global).
2. **Estacionalidad:** Patrones que se repiten en intervalos fijos (ej. ventas en Navidad).
3. **Ciclos:** Fluctuaciones que no tienen una periodicidad fija (ej. ciclos económicos).
4. **Ruido:** Variaciones aleatorias que no siguen un patrón claro.

### Ejemplo de Series Temporales
Vamos a ver algunos ejemplos de series temporales en diferentes contextos:
- **Clima:** Temperaturas diarias en una ciudad.
- **Economía:** Precios mensuales de una acción en bolsa.
- **Salud:** Número de casos diarios de una enfermedad.
- **Tecnología:** Tráfico de datos en una red por hora.

## 2. Importancia de las Series Temporales

Las series temporales son fundamentales en muchos campos, como:
- **Predicción:** Predecir valores futuros basados en datos históricos (ej. demanda de productos).
- **Monitoreo:** Detectar anomalías en tiempo real (ej. fraudes en transacciones bancarias).
- **Planificación:** Optimizar recursos basados en patrones históricos (ej. inventario en tiendas).
- **Investigación:** Analizar tendencias y patrones en datos científicos (ej. cambio climático).

## 3. Visualización de Series Temporales

La visualización es una herramienta clave para entender las series temporales. Vamos a cargar y visualizar algunos ejemplos utilizando Python y las bibliotecas `pandas` y `matplotlib`.

```python
import pandas as pd
import matplotlib.pyplot as plt

# Ejemplo 1: Temperatura mensual
data_temp = {
    'Mes': ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic'],
    'Temperatura': [12, 14, 16, 18, 22, 25, 28, 29, 26, 22, 18, 14]
}

df_temp = pd.DataFrame(data_temp)

plt.figure(figsize=(10, 6))
plt.plot(df_temp['Mes'], df_temp['Temperatura'], marker='o', label='Temperatura')
plt.title('Temperatura Mensual')
plt.xlabel('Mes')
plt.ylabel('Temperatura (°C)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Ejemplo 2: Ventas mensuales de un producto
data_ventas = {
    'Mes': ['Ene', 'Feb', 'Mar', 'Abr', 'May', 'Jun', 'Jul', 'Ago', 'Sep', 'Oct', 'Nov', 'Dic'],
    'Ventas': [120, 150, 130, 160, 200, 180, 220, 250, 210, 190, 170, 300]
}

df_ventas = pd.DataFrame(data_ventas)

plt.figure(figsize=(10, 6))
plt.plot(df_ventas['Mes'], df_ventas['Ventas'], marker='o', color='green', label='Ventas')
plt.title('Ventas Mensuales')
plt.xlabel('Mes')
plt.ylabel('Ventas (unidades)')
plt.grid(True)
plt.legend()
plt.show()
```

## 4. Componentes de una Serie Temporal

Las series temporales suelen descomponerse en varios componentes:
1. **Tendencia (Trend):** Cambio gradual a largo plazo.
2. **Estacionalidad (Seasonality):** Patrones que se repiten en intervalos fijos.
3. **Ciclos (Cycles):** Fluctuaciones que no tienen una periodicidad fija.
4. **Ruido (Noise):** Variaciones aleatorias.

### Ejemplo de Descomposición
Vamos a descomponer una serie temporal en sus componentes utilizando `statsmodels`.

```python
from statsmodels.tsa.seasonal import seasonal_decompose
import numpy as np

# Crear una serie temporal de ejemplo con tendencia y estacionalidad
np.random.seed(0)
time = np.arange(0, 200, 1)
trend = 0.1 * time
seasonality = 10 * np.sin(time / 5)
noise = np.random.normal(0, 2, len(time))
series = trend + seasonality + noise

# Descomponer la serie
decomposition = seasonal_decompose(series, period=10)

# Graficar los componentes
plt.figure(figsize=(12, 8))
decomposition.plot()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import statsmodels.api as sm
from statsmodels.tsa.seasonal import seasonal_decompose
from statsmodels.tsa.stattools import adfuller
import numpy as np

# 1. Cargar dataset de pasajeros aéreos
data = sm.datasets.get_rdataset("AirPassengers", "datasets").data
print("Dataset cargado:")
print(data.head())
print(f"\nDimensiones: {data.shape}")
print(f"Columnas: {data.columns.tolist()}")
```

```python
data['Month'] = pd.date_range(start='1949-01-01', periods=len(data), freq='M')
data = data.set_index('Month')
data = data[['value']]  # Seleccionar solo la columna value
data.columns = ['Pasajeros']  # Renombrar columna para claridad

print("Datos preparados con índice temporal:")
print(data.head())
print(f"\nPeriodo: {data.index.min()} a {data.index.max()}")
print(f"Total de observaciones: {len(data)}")
```

```python
plt.figure(figsize=(12, 6))
plt.plot(data.index, data['Pasajeros'], marker='o', linestyle='-', linewidth=2, markersize=4)
plt.title('Serie Temporal: Pasajeros Aéreos Mensuales (1949-1960)', fontsize=14, fontweight='bold')
plt.xlabel('Fecha', fontsize=12)
plt.ylabel('Número de Pasajeros', fontsize=12)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()

print("\nOBSERVACIONES INICIALES:")
print("- Tendencia: Se observa una tendencia creciente a lo largo del tiempo")
print("- Estacionalidad: Hay patrones que se repiten anualmente (picos y valles regulares)")
print("- Variabilidad: La amplitud de las oscilaciones aumenta con el tiempo")
```

```python
print("=" * 60)
print("DESCOMPOSICIÓN DE LA SERIE TEMPORAL")
print("=" * 60)

decomp = seasonal_decompose(data['Pasajeros'], model='multiplicative', period=12)

fig = decomp.plot()
fig.set_size_inches(12, 10)
plt.suptitle('Descomposición de la Serie Temporal (Modelo Multiplicativo)', fontsize=14, fontweight='bold', y=0.995)
plt.tight_layout()
plt.show()

print("\nANÁLISIS DE COMPONENTES:")
print("\n1. TENDENCIA (Trend):")
print("   - Muestra el crecimiento sostenido a largo plazo")
print("   - Línea suave sin las fluctuaciones estacionales")
print("\n2. ESTACIONALIDAD (Seasonal):")
print("   - Patrón que se repite cada 12 meses")
print("   - Picos en verano (vacaciones) y valles en invierno")
print("\n3. RESIDUO (Residual):")
print("   - Variaciones aleatorias no explicadas por tendencia ni estacionalidad")
print("   - Representan eventos inesperados o ruido aleatorio")
```

```python
print("=" * 60)
print("TEST DE ESTACIONARIEDAD (AUGMENTED DICKEY-FULLER)")
print("=" * 60)

result = adfuller(data['Pasajeros'])

print(f"\nEstadístico ADF: {result[0]:.4f}")
print(f"p-value: {result[1]:.4f}")
print(f"\nNúmero de lags usados: {result[2]}")
print(f"Número de observaciones: {result[3]}")

print("\nValores críticos:")
for key, value in result[4].items():
    print(f"   {key}: {value:.3f}")

print("\n" + "=" * 60)
print("INTERPRETACIÓN:")
print("=" * 60)

if result[1] > 0.05:
    print("\n❌ La serie NO ES ESTACIONARIA")
    print(f"   - p-value ({result[1]:.4f}) > 0.05")
    print("   - No se rechaza la hipótesis nula (existe raíz unitaria)")
    print("\nRAZONES:")
    print("   1. La media cambia con el tiempo (tendencia creciente)")
    print("   2. La varianza aumenta a medida que pasan los años")
    print("   3. Hay estacionalidad (patrones repetitivos)")
else:
    print("\n✅ La serie ES ESTACIONARIA")
    print(f"   - p-value ({result[1]:.4f}) <= 0.05")
    print("   - Se rechaza la hipótesis nula")

print("\nRECOMENDACIONES:")
print("   - Aplicar diferenciación para eliminar la tendencia")
print("   - Considerar transformación logarítmica para estabilizar varianza")
print("   - Usar modelos ARIMA o SARIMA para predicción")
```

```python
print("=" * 60)
print("TRANSFORMACIONES PARA LOGRAR ESTACIONARIEDAD")
print("=" * 60)

# Transformación logarítmica para estabilizar varianza
data['log_pasajeros'] = np.log(data['Pasajeros'])

# Diferenciación para eliminar tendencia
data['diff_log_pasajeros'] = data['log_pasajeros'].diff()

# Visualizar las transformaciones
fig, axes = plt.subplots(3, 1, figsize=(12, 10))

# Serie original
axes[0].plot(data.index, data['Pasajeros'])
axes[0].set_title('Serie Original', fontweight='bold')
axes[0].set_ylabel('Pasajeros')
axes[0].grid(True, alpha=0.3)

# Serie con transformación logarítmica
axes[1].plot(data.index, data['log_pasajeros'], color='orange')
axes[1].set_title('Transformación Logarítmica', fontweight='bold')
axes[1].set_ylabel('log(Pasajeros)')
axes[1].grid(True, alpha=0.3)

# Serie diferenciada (más estacionaria)
axes[2].plot(data.index, data['diff_log_pasajeros'], color='green')
axes[2].set_title('Diferenciación de log(Pasajeros)', fontweight='bold')
axes[2].set_ylabel('Δ log(Pasajeros)')
axes[2].set_xlabel('Fecha')
axes[2].grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print("\nEXPLICACIÓN:")
print("1. Serie original: Varianza creciente y tendencia clara")
print("2. Log(Pasajeros): Estabiliza la varianza (oscilaciones más uniformes)")
print("3. Diferenciación: Elimina la tendencia, oscila alrededor de cero")
print("\nLa serie diferenciada está más cerca de ser estacionaria.")
```
