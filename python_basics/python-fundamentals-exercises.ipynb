## Ejercicio 1: Calculadora de Estado de un Robot

**Enunciado:**
1.  Crea dos variables: `nivel_bateria` con un valor entero (ej: `75`) y `tarea_actual` con un valor de texto (ej: `"limpiando"`).
2.  Escribe una estructura condicional (`if`/`elif`/`else`) que imprima un mensaje diferente según el nivel de la batería:
    - Si la batería es **menor o igual a 20%**: `"Batería crítica ({nivel_bateria}%) - ¡Volviendo a la base de carga!"`
    - Si la batería está **entre 21% y 70%**: `"Batería estable ({nivel_bateria}%) - Realizando tarea: {tarea_actual}."`
    - Si la batería es **mayor de 70%**: `"Batería óptima ({nivel_bateria}%) - Máximo rendimiento en tarea: {tarea_actual}."`
3.  Prueba a cambiar los valores de las variables y vuelve a ejecutar la celda para ver cómo cambia el mensaje.

### Tu Explicación (Ejercicio 1)

*(Haz doble clic aquí para escribir tu explicación haciendo uso de Markdown.)

```python
nivel_bateria = 75
tarea_actual = "limpiando"

if nivel_bateria <= 20:
    print(f"Batería crítica ({nivel_bateria}%) - ¡Volviendo a la base de carga!")
elif 21 <= nivel_bateria <= 70:
    print(f"Batería estable ({nivel_bateria}%) - Realizando tarea: {tarea_actual}.")
else:
    print(f"Batería óptima ({nivel_bateria}%) - Máximo rendimiento en tarea: {tarea_actual}.")
```

**Enunciado:**
1.  Crea una **lista** de strings llamada `ordenes_robot` que contenga al menos cuatro comandos (ej: `"avanzar"`, `"escanear_perimetro"`, `"girar_derecha"`, `"recoger_muestra"`).
2.  Utiliza un **bucle `for`** para recorrer la lista.

## Ejercicio 4: Función de Triaje Simple

**Enunciado:**
1.  Crea una **función** llamada `evaluar_paciente` que acepte dos argumentos: `temperatura` (un float) y `saturacion_oxigeno` (un int).
2.  La función debe seguir estas reglas para **devolver** (`return`) un string con el diagnóstico:
    - Si la `temperatura` es mayor de 37.5 **y** la `saturacion_oxigeno` es menor de 95, debe devolver `"Atención urgente"`.
    - Si solo la `temperatura` es mayor de 37.5, debe devolver `"Observación por fiebre"`.
    - Si solo la `saturacion_oxigeno` es menor de 95, debe devolver `"Revisar saturación"`.
    - En cualquier otro caso, debe devolver `"Sin riesgo aparente"`.
3.  Llama a la función varias veces con diferentes valores para probar todas las combinaciones posibles y muestra los resultados con `print`.

### Tu Explicación (Ejercicio 4)

*(Haz doble clic aquí para escribir tu explicación haciendo uso de Markdown.)

```python
def evaluar_paciente(temperatura, saturacion_oxigeno):
    if temperatura > 37.5 and saturacion_oxigeno < 95:
        return "Atención urgente"
    elif temperatura > 37.5:
        return "Observación por fiebre"
    elif saturacion_oxigeno < 95:
        return "Revisar saturación"
    else:
        return "Sin riesgo aparente"

print(evaluar_paciente(38.0, 93))
print(evaluar_paciente(38.5, 96))
print(evaluar_paciente(36.0, 92))
print(evaluar_paciente(36.8, 97))
```

## Ejercicio 5: Chequeo de Sensores (Variables, Operadores y Condicionales)

**Enunciado:**
Un robot de exploración necesita verificar que sus sensores operan en un rango seguro.
1.  Crea dos variables: `temperatura_cpu` (ej: `65.5`) y `carga_bateria` (ej: `80`).
2.  Escribe una estructura condicional que imprima el estado del robot basado en estas reglas, usando **operadores lógicos** (`and`, `or`):
    - Si la `temperatura_cpu` es **mayor a 85** o la `carga_bateria` es **menor a 20**, imprime: `"🔴 ALERTA CRÍTICA: Parando operaciones."`
    - Si la `temperatura_cpu` está **entre 60 y 85** (inclusives) **y** la `carga_bateria` es **mayor o igual a 20**, imprime: `"🟡 ESTADO: Óptimo. Continuando misión."`
    - En cualquier otro caso (temperatura por debajo de 60), imprime: `"🔵 ESTADO: En reposo. Temperatura baja."`
3.  Prueba con diferentes valores para activar todas las condiciones.

### Tu Explicación (Ejercicio 5)

*(Haz doble clic aquí para escribir tu explicación haciendo uso de Markdown.)

```python
temperatura_cpu = 65.5
carga_bateria = 80

if temperatura_cpu > 85 or carga_bateria < 20:
    print("🔴 ALERTA CRÍTICA: Parando operaciones.")
elif 60 <= temperatura_cpu <= 85 and carga_bateria >= 20:
    print("🟡 ESTADO: Óptimo. Continuando misión.")
else:
    print("🔵 ESTADO: En reposo. Temperatura baja.")
```

## Ejercicio 6: Inventario de Componentes (Listas y Tuplas)

**Enunciado:**
Necesitas gestionar un inventario de componentes de hardware para un sistema de IA. Cada componente tiene datos que no deberían cambiar: un ID de producto y una fecha de fabricación.
1.  Crea al menos tres **tuplas**, donde cada una represente un componente con `(id_producto, nombre, fecha_fabricacion)`. Por ejemplo: `("NVD-001", "GPU A100", "2023-05-12")`.

Puedes usar los siguientes datos como ejemplo:
* "NVD-001", "GPU A100", "2023-05-12"
* "INT-C01", "CPU Xeon Gold 6248R", "2023-03-20"
* "SAM-M01", "Memoria RAM 128GB DDR4 ECC", "2023-01-15"

## Ejercicio 7: Análisis de Etiquetas de Imágenes (Sets y Diccionarios)

**Enunciado:**
Un modelo de visión por computador ha procesado un lote de imágenes y ha devuelto una lista de etiquetas detectadas. La lista contiene duplicados.
1.  Crea una **lista** llamada `etiquetas_detectadas` con varias etiquetas, algunas repetidas. Ej: `"algoritmo", "datos", "modelo", "python", "datos", "algoritmo", "tensor", "modelo", "aprendizaje", "datos"`.
2.  Usa un **set** para obtener una colección de las etiquetas **únicas** que el modelo ha sido capaz de identificar. Imprime las etiquetas únicas.
3.  Crea un **diccionario** llamado `frecuencia_etiquetas` que cuente cuántas veces aparece cada etiqueta en la lista original. Imprime el diccionario.

### Tu Explicación (Ejercicio 7)

*(Haz doble clic aquí para escribir tu explicación haciendo uso de Markdown.)

```python
etiquetas_detectadas = [
    "algoritmo", "datos", "modelo", "python", "datos", "algoritmo",
    "tensor", "modelo", "aprendizaje", "datos"
]

etiquetas_unicas = set(etiquetas_detectadas)
print("Etiquetas únicas:", etiquetas_unicas)

frecuencia_etiquetas = {}
for etiqueta in etiquetas_detectadas:
    frecuencia_etiquetas[etiqueta] = frecuencia_etiquetas.get(etiqueta, 0) + 1
print("Frecuencia de etiquetas:", frecuencia_etiquetas)
```

## Ejercicio 8: Generador de Perfiles (Importación y Funciones)

**Enunciado:**
Necesitas una función para generar datos de prueba aleatorios para un sistema experto.
1.  **Importa** la librería `random` y dale el alias `rnd`.
2.  Crea una **función** llamada `generar_perfil_aleatorio()` que no reciba argumentos.
3.  Dentro de la función, usa `rnd.choice()` para seleccionar un diagnóstico aleatorio de una lista de posibilidades (ej: `["Sano", "Fiebre Leve", "Observación"]`) y `rnd.randint()` para generar una edad aleatoria entre 18 y 80.
4.  La función debe **devolver** un diccionario con la edad y el diagnóstico generados.
5.  Llama a la función e imprime el perfil aleatorio que ha generado.

### Tu Explicación (Ejercicio 8)

*(Haz doble clic aquí para escribir tu explicación haciendo uso de Markdown.)

```python
import random as rnd

def generar_perfil_aleatorio():
    diagnostico = rnd.choice(["Sano", "Fiebre Leve", "Observación"])
    edad = rnd.randint(18, 80)
    return {"edad": edad, "diagnóstico": diagnostico}

perfil = generar_perfil_aleatorio()
print("Perfil generado:", perfil)
```

## Ejercicio 9: Normalización de Datos.

En Machine Learning, es una práctica muy común normalizar los datos para que estén en un rango específico, normalmente entre 0 y 1. Imagina que tienes una lista con los valores de los píxeles de una imagen en escala de grises (que van de 0 a 255).

Crea una lista llamada valores_pixeles con algunos valores enteros entre 0 y 255. Por ejemplo: [15, 80, 128, 210, 255].

Usando una única list comprehension, crea una nueva lista llamada pixeles_normalizados donde cada valor de la lista original haya sido dividido por 255.0 para escalarlo entre 0 y 1.

Imprime la lista pixeles_normalizados para ver el resultado.

### Tu Explicación (Ejercicio 9)

*(Haz doble clic aquí para escribir tu explicación haciendo uso de Markdown.)

```python
valores_pixeles = [15, 80, 128, 210, 255]
pixeles_normalizados = [valor / 255.0 for valor in valores_pixeles]
print("Pixeles normalizados:", pixeles_normalizados)
```

## Ejercicio 10: Normalización de Datos.

En el Procesamiento del Lenguaje Natural (PLN), un primer paso habitual es eliminar las "palabras vacías" (stop words), que son palabras muy comunes pero con poco significado (el, y, de, los, etc.).

Te damos una frase representada como una lista de palabras: frase = ["el", "sistema", "experto", "analiza", "los", "datos", "con", "precisión"].

Te damos también una lista de stop words: stop_words = ["el", "los", "las", "un", "una", "de", "con", "en"].

Usando una única list comprehension, crea una nueva lista llamada frase_filtrada que contenga únicamente las palabras de frase que no están en la lista stop_words.

Imprime la frase_filtrada.

### Tu Explicación (Ejercicio 10)

*(Haz doble clic aquí para escribir tu explicación haciendo uso de Markdown.)

```python
frase = ["el", "sistema", "experto", "analiza", "los", "datos", "con", "precisión"]
stop_words = ["el", "los", "las", "un", "una", "de", "con", "en"]

frase_filtrada = [palabra for palabra in frase if palabra not in stop_words]
print("Frase filtrada:", frase_filtrada)
```
