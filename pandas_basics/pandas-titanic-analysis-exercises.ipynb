### Parte 3: Manipulación y Limpieza de Datos

#### Ejercicio 8: Creación de una Nueva Columna

#### Ejercicio 9: Tratamiento de Valores Nulos
**Enunciado:**
1. Primero, calcula y muestra cuántos valores nulos hay en la columna `age`.
2. Rellena los valores nulos de la columna `age` con la **mediana** de edad de todos los pasajeros.
3. Verifica que ya no quedan valores nulos en esa columna.

#### Tu Explicación (Ejercicio 9)
*(Escribe aquí tu explicación...)*

```python
n_nulos = df['age'].isnull().sum()
print("Valores nulos en 'age':", n_nulos)
mediana_edad = df['age'].median()
print("Mediana de edad:", mediana_edad)
df['age'] = df['age'].fillna(mediana_edad)
n_nulos_restantes = df['age'].isnull().sum()
print("Valores nulos en 'age' tras rellenar:", n_nulos_restantes)
```

### Parte 4: Agregación y Análisis - Extrayendo Conocimiento

#### Ejercicio 11: Agrupación Simple con `.groupby()`
**Enunciado:** Calcula la **tarifa media (`fare`)** que pagaron los pasajeros de cada clase (`pclass`).

#### Tu Explicación (Ejercicio 11)

#### Ejercicio 12: Agrupación Múltiple con `.agg()`
**Enunciado:** Agrupa el DataFrame por **clase (`pclass`)** y **sexo (`sex`)**. Para cada grupo, calcula:
- El número total de supervivientes.
- La edad media.
- La tarifa máxima pagada.

#### Tu Explicación (Ejercicio 12)
*(Escribe aquí tu explicación...)*

```python
df.groupby(['pclass', 'sex']).agg({
    'survived': 'sum',
    'age': 'mean',
    'fare': 'max'
})
```
